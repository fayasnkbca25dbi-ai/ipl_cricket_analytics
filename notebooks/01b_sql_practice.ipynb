```python
# Part B - Question 1
q("""
SELECT *
FROM matches
WHERE season = 2019;
""")

# My observation:
# WHERE filters the matches and shows only matches from the 2019 season.
```

```python
# Part B - Question 2
q("""
SELECT *
FROM matches
WHERE win_by_runs > 100;
""")

# My observation:
# This query shows matches where the winning margin in runs is greater than 100.
```

```python
# Part B - Question 3
q("""
SELECT *
FROM matches
WHERE season = 2016 AND win_by_wickets > 5;
""")

# My observation:
# AND applies both conditions and shows 2016 matches won by more than 5 wickets.
```

```python
# Part B - Question 4
q("""
SELECT match_id, win_by_runs
FROM matches
ORDER BY win_by_runs DESC
LIMIT 10;
""")

# My observation:
# DESC sorts the largest winning margins first and LIMIT returns the top 10.
```

```python
# Part B - Question 5
q("""
SELECT match_id, win_by_wickets
FROM matches
WHERE win_by_wickets IS NOT NULL
ORDER BY win_by_wickets ASC
LIMIT 10;
""")

# My observation:
# IS NOT NULL removes missing values and ASC sorts the smallest wicket margins first.
```

```python
# Part C - Question 1
q("""
SELECT COUNT(*) AS total_matches
FROM matches;
""")

# My observation:
# COUNT(*) counts the total number of matches.
```

```python
# Part C - Question 2
q("""
SELECT COUNT(*) AS total_deliveries
FROM deliveries;
""")

# My observation:
# COUNT(*) counts all delivery records in the deliveries table.
```

```python
# Part C - Question 3
q("""
SELECT COUNT(DISTINCT venue) AS total_venues
FROM matches;
""")

# My observation:
# COUNT(DISTINCT venue) counts each venue only once.
```

```python
# Part C - Question 4
q("""
SELECT venue, COUNT(*) AS matches_hosted
FROM matches
GROUP BY venue
ORDER BY matches_hosted DESC;
""")

# My observation:
# GROUP BY groups matches by venue and shows the venues with the most matches first.
```

```python
# Part C - Question 5
q("""
SELECT match_winner, COUNT(*) AS wins
FROM matches
GROUP BY match_winner
ORDER BY wins DESC;
""")

# My observation:
# This query counts the number of wins for each match winner and sorts them highest first.
```

```python
# Part C - Question 6
q("""
SELECT venue, COUNT(*) AS matches_hosted
FROM matches
GROUP BY venue
HAVING COUNT(*) >= 20
ORDER BY matches_hosted DESC;
""")

# My observation:
# HAVING filters grouped results and shows only venues with at least 20 matches.
```

```python
# Part C - Question 7
q("""
SELECT bowler, COUNT(*) AS deliveries_bowled
FROM deliveries
GROUP BY bowler
HAVING COUNT(*) > 2000
ORDER BY deliveries_bowled DESC;
""")

# My observation:
# This query counts deliveries for each bowler and shows bowlers with more than 2,000 deliveries.
```

```python
# Part D - Question 1
q("""
SELECT COUNT(*) AS missing_city_matches
FROM matches
WHERE city IS NULL;
""")

# My observation:
# IS NULL finds matches where the city value is missing.
```

```python
# Part D - Question 2
q("""
SELECT COUNT(*) AS missing_match_number
FROM matches
WHERE match_number IS NULL;
""")

# My observation:
# This query counts matches where match_number has no value.
```

```python
# Part D - Question 3
q("""
SELECT field_pos, LENGTH(field_pos) AS field_pos_length
FROM players
WHERE field_pos IS NOT NULL
ORDER BY field_pos_length DESC;
""")

# My observation:
# LENGTH shows the number of characters in field_pos and helps identify unusually long or suspicious values.
```

= NULL is not the correct test because NULL represents a missing or unknown value.
SQL uses IS NULL to check for NULL values.

```python
# Part E - Question 1
q("""
SELECT m.match_id, m.venue, v.city
FROM matches AS m
JOIN venues AS v
ON m.venue = v.venue;
""")

# My observation:
# JOIN combines matching venue records from the matches and venues tables.
```

```python
# Part E - Question 2
q("""
SELECT d.batter, p.bat_style
FROM deliveries AS d
JOIN players AS p
ON d.batter = p.player_name;
""")

# My observation:
# JOIN connects each batter with the matching player's batting style.
```

```python
# Part E - Question 3
q("""
SELECT DISTINCT d.batter
FROM deliveries AS d
LEFT JOIN players AS p
ON d.batter = p.player_name
WHERE p.player_name IS NULL;
""")

# My observation:
# LEFT JOIN with IS NULL finds batters that do not have a matching record in players.
```

```python
# Part E - Question 4
q("""
SELECT DISTINCT m.venue
FROM matches AS m
LEFT JOIN venues AS v
ON m.venue = v.venue
WHERE v.venue IS NULL;
""")

# My observation:
# This query finds match venues that do not have a matching record in the venues table.
```

```python
# Part F - Question 1
q("""
SELECT match_id, city,
       CASE
           WHEN city IS NULL THEN 'Yes'
           ELSE 'No'
       END AS city_missing
FROM matches;
""")

# My observation:
# CASE WHEN creates city_missing and marks whether the city value is missing.
```

```python
# Part F - Question 2
q("""
SELECT COUNT(*) AS total_matches,
       SUM(CASE WHEN city IS NULL THEN 1 ELSE 0 END) AS missing_city_matches
FROM matches;
""")

# My observation:
# This query shows the total number of matches and the number of matches with a missing city in one result.
```

```python
# Part F - Question 3
q("""
SELECT venue
FROM venues

EXCEPT

SELECT venue
FROM matches;
""")

# My observation:
# EXCEPT finds venues that exist in the venues table but never appear in the matches table.
```

```python
# Part F - Question 4
q("""
SELECT venue
FROM matches

EXCEPT

SELECT venue
FROM venues;
""")

# My observation:
# Reversing EXCEPT finds venues used in matches that are missing from the venues table.
```
